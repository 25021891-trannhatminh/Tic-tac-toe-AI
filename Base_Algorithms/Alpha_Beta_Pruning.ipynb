```python
root = TreeNode()

A = TreeNode()
A.add_child(TreeNode(3))
A.add_child(TreeNode(12))
A.add_child(TreeNode(8))

B = TreeNode()
B.add_child(TreeNode(2))
B.add_child(TreeNode(4))
B.add_child(TreeNode(6))

C = TreeNode()
C.add_child(TreeNode(14))
C.add_child(TreeNode(5))
C.add_child(TreeNode(2))

root.add_child(A)
root.add_child(B)
root.add_child(C)
```

```python
def Alpha_Beta_Pruning(node,is_max,alpha = float('-inf'),beta = float('inf')):
  if len(node.children) ==0:
    return node.value

  if is_max:
    best = float('-inf')
    for child in node.children:
      value = Alpha_Beta_Pruning(child,False,alpha,beta)
      best = max(value,best)
      alpha = max(best,alpha)

      #Alpha-Pruning
      if alpha >= beta:
        break
    return best

  else:
    best = float('inf')
    for child in node.children:
      value = Alpha_Beta_Pruning(child,True,alpha,beta)
      best = min(best,value)
      beta = min(best,beta)

      #Beta-Pruning
      if alpha >= beta:
        break
    return best
```
