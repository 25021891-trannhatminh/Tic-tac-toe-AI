

Example Tree

```python
class TreeNode:
    def __init__(self,value= None):
        self.value  = value 
        self.children = []
    def add_child(self,child):
        self.children.append(child)
```

```python
#Build Tree
root = TreeNode()

A = TreeNode()
A.add_child(TreeNode(3))
A.add_child(TreeNode(12))
A.add_child(TreeNode(8))

B = TreeNode()
B.add_child(TreeNode(2))
B.add_child(TreeNode(4))
B.add_child(TreeNode(6))

C = TreeNode()
C.add_child(TreeNode(14))
C.add_child(TreeNode(5))
C.add_child(TreeNode(2))

root.add_child(A)
root.add_child(B)
root.add_child(C)
```

```python
def MiniMax_TreeNode(node, is_max):
    if len(node.children) ==0:
        return node.value 

    if is_max:
        best = float('-inf')
        for child in node.children:
            value = MiniMax_TreeNode(child,False)
            best = max(best,value)
    else:
        best = float('inf')
        for child in node.children:
            value = MiniMax_TreeNode(child,True)
            best = min(best,value)

    return best 
```

```python
MiniMax_TreeNode(root,True)
```
