Rules: (Game = Tic Tac Toe)
+ Player X: 1
+ Player 0: -1
+ Blank : 0
  
Reward - wins: according to the parent node - NOT THE CURRENT NODE
+ +1 : win


```python
import math 
import random 
import numpy as np 
```

```python
class Game:
  SIZE = 10
  WIN_LENGTH =5
  @staticmethod
  def legal_actions(state):
    board,player = state
    return [
        (row, col)
        for row in range(Game.SIZE)
        for col in range(Game.SIZE)
        if board[row][col] == 0
    ]

  @staticmethod
  def next_state(state,action): # action = legit chose blank for the current player to take
    board,player = state
    row, col = action

    new_board = [list(r) for r in board]

    new_board[row][col] = player

    new_board = tuple(tuple(r) for r in new_board)

    return (new_board,-player)

  @staticmethod
  def get_winner(state):
    board , _ = state
    directions = [
    #(row, column)
        (0,1),
        (1,0),
        (1,1),
        (-1,1)
    ]
    for row in range(Game.SIZE):
      for column in range(Game.SIZE):
        player = board[row][column]
        if player ==0:
          continue
        for dr, dc in directions:
          count = 0
          for k in range(Game.WIN_LENGTH):
            nr = row + dr *k
            nc = column + dc *k

            if not(0<=nr<Game.SIZE and 0<= nc< Game.SIZE):
              break

            if board[nr][nc] == player:
              count +=1
            else:
              break

          if count == Game.WIN_LENGTH:
            return player
    return None

  @staticmethod
  def is_terminal(state):
    board, _ = state
    has_winner = Game.get_winner(state) is not None
    board_full = all(
        cell != 0
        for row in board
        for cell in row
    )

    return has_winner or board_full



  @staticmethod
  def rollout(state):
    current_state = state
    rollout_count = 0
    while not Game.is_terminal(current_state):
      valid_actions = Game.legal_actions(current_state)
      action = random.choice(valid_actions)
      current_state = Game.next_state(current_state,action)
      rollout_count +=1
      winner = Game.get_winner(current_state)

      print("Moves_ROLLOUT:", rollout_count)

    return winner
```

```python
#Test: get_winner()

#Test 1:
board_1 = [[0] * 10 for _ in range(10)]
for col in range(5):
    board_1[2][col] = 1
board_1_new = tuple(tuple(row) for row in board_1)
state_1 = (board_1_new, -1)
winner_1 = Game.get_winner(state_1)
if winner_1 ==1 :
  print("TEST 1 PASSED")

#Test 2:
board_2 = [[0] * 10 for _ in range(10)]
for i in range(5):
    board_2[i][i] = -1
board_2_new = tuple(tuple(row) for row in board_2)
state_2 = (board_2_new, 1)
winner_2 = Game.get_winner(state_2)
if winner_2 == -1 :
  print("TEST 2 PASSED")

# Test 3: Vertical
board_3 = [[0] * 10 for _ in range(10)]
for row in range(5):
    board_3[row][3] = 1

board_3_new = (tuple(tuple(row) for row in board_3))
state_3 = (board_3_new,-1)
winner_3 = Game.get_winner(state_3)
if winner_3 ==1 :
    print("TEST 3 PASSED")


# Test 4: Anti-diagonal
board_4 = [[0] * 10 for _ in range(10)]
for i in range(5):
    board_4[i][4-i] = -1

board_4_new = (tuple(tuple(r) for r in board_4))
state_4 = (board_4_new,1)
winner_4 = Game.get_winner(state_4)
if winner_4 == -1:
    print("TEST 4 PASSED")
```
