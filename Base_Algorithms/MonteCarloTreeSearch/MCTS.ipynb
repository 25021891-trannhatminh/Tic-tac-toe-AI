```python
from ipynb.fs.defs.Game import Game 
import math 
import random
import numpy as np 
```

```python
class Node: 
    def __init__(self,state,parent = None,action = None): # action: action that Parent(Current Node) takes to reach Current Node 
        self.state = state 
        self.parent = parent 
        self.action = action 

        self.children = []

        self.visits = 0 
        self.rewards = 0.0 

        self.untried_actions = Game.legal_actions(state)

        self.player = state[1]

    def is_fully_expanded(self):
        return len(self.untried_actions)==0
    def is_terminal(self):
        return Game.is_terminal(self.state)
```

```python
# UCT Formular
def UCT(parent,child,Constant = math.sqrt(2)):
  exploitation = child.rewards/child.visits
  exploration = Constant * math.sqrt( math.log(parent.visits)/ child.visits )
  UCT_RES = exploitation + exploration
  return UCT_RES
```

```python
#FORWARD ITERTAION:
    #Start iteration: Node_Selection ( Tree Traversal)
    #-->Node_Expansion 
    #-->Rollout ( Simulation )
    #-->Backpropagation --> End iteration
```

```python
#Node Selection
def selection(node):
  while not node.is_terminal():
    if not node.is_fully_expanded():
      return node

    node = max(node.children, key = lambda child : UCT(node, child))
  return node
```

```python
# Node expansion
def expansion(node):
  if len(node.untried_actions)==0:
    return node
  
  action = random.choice(node.untried_actions)
  node.untried_actions.remove(action)
  new_state = Game.next_state(node.state,action)
  
  child = Node(
      state = new_state,
      parent = node,
      action = action
  )
  node.children.append(child)
  return child
```

```python
# Simulation ( Rollout)
def rollout(node):
  return Game.rollout(node.state)
```

```python
# Backpropagation
def backpropagation(node, winner):
  while node is not None:
    node.visits +=1
    if winner is None:
      node.rewards += 0.5

    elif node.parent is None: #Root
      if winner == node.player:
        node.rewards +=1
    else:
      if winner == node.parent.player:
        node.rewards +=1
    node = node.parent
```

```python
# Combination: Forward iteration
def forward_iteration(root):
  node = selection(root)
  print("Selected Node:", node)
  print("Selected Action:", node.action)

  if node.is_terminal():
    winner= Game.get_winner(node.state)
  else:
    node = expansion(node)
    print("Expanded action:", node.action)
    winner = rollout(node)
    print("Winner:", winner)

  backpropagation(node,winner)
```

```python
# Checking function:
def print_tree(node, depth=0):

    indent = "  " * depth

    print(
        indent,
        f"Action={node.action}, "
        f"Player={node.player}, "
        f"Visits={node.visits}, "
        f"Rewards={node.rewards}, "
        f"Untried={len(node.untried_actions)}"
    )

    for child in node.children:
        print_tree(child, depth + 1)
```

```python
#Test
initial_board = tuple(
    tuple([0] * 10)
    for _ in range(10)
)

initial_state = (initial_board, 1)

root = Node(initial_state)
```

```python
for i in range(100):
    print(f"\n========== ITERATION {i + 1} ==========")

    forward_iteration(root)

    print_tree(root)

print("\n========== FINAL RESULT ==========")

print("Root visits:", root.visits)
print("Number of children:", len(root.children))

for child in root.children:
    print(
        f"Action={child.action}, "
        f"Visits={child.visits}, "
        f"Wins={child.rewards}"
    )
```
